```python
from pathlib import Path
import hashlib, json, torch
root = Path('/workspace')
checkpoint = root / 'models/lightweight_vla_adapter/universal_three_scene_v6_sensor_policy_finetuned_stage8/model.pt'
config = root / 'LMM-in-AutoDrive/lightweight_vla_adapter/configs/universal_three_scene_v6_sensor_policy.json'
agent = root / 'Bench2Drive/leaderboard/team_code/universal_vla_agent.py'
assert checkpoint.is_file() and config.is_file() and agent.is_file()
print('torch', torch.__version__, 'cuda', torch.cuda.is_available())
```

```python
digest = hashlib.sha256(checkpoint.read_bytes()).hexdigest()
expected = '53e949b37c84d6010ab45bfd473cb9d39a88cd89cd7729f55d3e9bb1baddaad3'
assert digest == expected, (digest, expected)
print('checkpoint SHA256 OK:', digest)
```
